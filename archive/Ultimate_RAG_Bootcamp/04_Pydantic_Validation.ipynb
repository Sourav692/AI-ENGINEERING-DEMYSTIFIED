### Pydantic Data Validation

```python
from langgraph.graph import StateGraph,START,END
from pydantic import BaseModel
from typing import Annotated
from langgraph.graph.message import add_messages
```

```python
class State(BaseModel):
    name:Annotated[list,add_messages]
    # name:str
```

```python
## node function
def example_node(state:State):
    state.name="Hello"
    return {"name":state.name}

def greeting(state:State):
    state.name="How are you?"
    return {"name":state.name}
```

```python
## stateGraph
builder=StateGraph(State)
builder.add_node("example_node",example_node)
builder.add_node("greeting",greeting)

builder.add_edge(START,"example_node")
builder.add_edge("example_node","greeting")
builder.add_edge("example_node",END)

graph=builder.compile()
```

```python
graph.invoke({"name":"Sourav"})
```

```python
graph.invoke({"name":"Hi"})
```

```python
graph.invoke({"name":123})
```
