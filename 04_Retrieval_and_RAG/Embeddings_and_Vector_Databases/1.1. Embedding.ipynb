## Measuring Similarity

```python
def cosine_similarity(vec1, vec2):
    """
    Cosine similarity measures the angle between two vectors.
    - Result close to 1: Very similar
    - Result close to 0: Not related
    - Result close to -1: Opposite meanings
    """

    dot_product=np.dot(vec1,vec2)
    norm_a=np.linalg.norm(vec1)
    norm_b=np.linalg.norm(vec2)
    return dot_product/(norm_a * norm_b)
```

```python
# Example
cat_vector = [0.8, 0.6, 0.3]
kitten_vector = [0.75, 0.65, 0.35]
car_vector = [-0.5, 0.2, 0.1]

cat_kitten_similarity=cosine_similarity(cat_vector,kitten_vector)
print(cat_kitten_similarity)
```

```python
cosine_similarity(cat_vector,car_vector)
```

### Creating Your First Embeddings

```python
### Huggingface And OpenAI Models

from langchain_huggingface import HuggingFaceEmbeddings

## Initialize a simple Embedding model(no API Key needed!)
embeddings=HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
embeddings
```

```python
## create your first embeddings
text="Hello, I am learning about embeddings!"

embedding=embeddings.embed_query(text)
print(f"Text: {text}")
print(f"Embedding length : {len(embedding)}")
print(embedding)
```

```python
embedding=embeddings.embed_documents(text)
print(f"Text: {text}")
print(f"Embedding length : {len(embedding)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
from matplotlib.patches import FancyArrowPatch, Arc

def cosine_similarity(v1, v2):
    """Calculate cosine similarity between two vectors"""
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    return dot_product / (norm_v1 * norm_v2)

def update(val):
    """Update the plot when sliders change"""
    # Get current slider values
    x1, y1 = slider_x1.val, slider_y1.val
    x2, y2 = slider_x2.val, slider_y2.val
    
    v1 = np.array([x1, y1])
    v2 = np.array([x2, y2])
    
    # Calculate cosine similarity and angle
    cos_sim = cosine_similarity(v1, v2)
    angle = np.arccos(np.clip(cos_sim, -1, 1)) * 180 / np.pi
    
    # Clear and redraw
    ax.clear()
    
    # Set up the plot
    max_range = 5
    ax.set_xlim(-max_range, max_range)
    ax.set_ylim(-max_range, max_range)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    
    # Draw vectors
    ax.arrow(0, 0, x1, y1, head_width=0.3, head_length=0.3, 
             fc='blue', ec='blue', linewidth=2, label='Vector 1')
    ax.arrow(0, 0, x2, y2, head_width=0.3, head_length=0.3, 
             fc='red', ec='red', linewidth=2, label='Vector 2')
    
    # Draw angle arc
    if angle > 0 and angle < 180:
        arc_radius = 1.5
        arc = Arc((0, 0), 2*arc_radius, 2*arc_radius, 
                  angle=0, theta1=min(np.arctan2(y1, x1), np.arctan2(y2, x2)) * 180/np.pi,
                  theta2=max(np.arctan2(y1, x1), np.arctan2(y2, x2)) * 180/np.pi,
                  color='green', linewidth=2)
        ax.add_patch(arc)
    
    # Add text information
    info_text = f'Vector 1: ({x1:.2f}, {y1:.2f})\n'
    info_text += f'Vector 2: ({x2:.2f}, {y2:.2f})\n\n'
    info_text += f'Cosine Similarity: {cos_sim:.4f}\n'
    info_text += f'Angle: {angle:.2f}°\n\n'
    info_text += f'Interpretation:\n'
    
    if cos_sim > 0.9:
        info_text += 'Very similar (nearly parallel)'
    elif cos_sim > 0.5:
        info_text += 'Moderately similar'
    elif cos_sim > 0:
        info_text += 'Somewhat similar'
    elif cos_sim > -0.5:
        info_text += 'Somewhat dissimilar'
    elif cos_sim > -0.9:
        info_text += 'Very dissimilar'
    else:
        info_text += 'Opposite direction'
    
    ax.text(0.02, 0.98, info_text, transform=ax.transAxes, 
            fontsize=10, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    ax.set_xlabel('X-axis')
    ax.set_ylabel('Y-axis')
    ax.set_title('Interactive Cosine Similarity Visualization', fontsize=14, fontweight='bold')
    ax.legend(loc='upper right')
    
    plt.draw()

# Create the figure and axis
fig, ax = plt.subplots(figsize=(12, 9))
plt.subplots_adjust(left=0.1, bottom=0.3)

# Initial vectors
init_x1, init_y1 = 3, 2
init_x2, init_y2 = 2, 4

# Create sliders
ax_x1 = plt.axes([0.15, 0.20, 0.7, 0.03])
ax_y1 = plt.axes([0.15, 0.15, 0.7, 0.03])
ax_x2 = plt.axes([0.15, 0.10, 0.7, 0.03])
ax_y2 = plt.axes([0.15, 0.05, 0.7, 0.03])

slider_x1 = Slider(ax_x1, 'V1 X', -5, 5, valinit=init_x1, valstep=0.1)
slider_y1 = Slider(ax_y1, 'V1 Y', -5, 5, valinit=init_y1, valstep=0.1)
slider_x2 = Slider(ax_x2, 'V2 X', -5, 5, valinit=init_x2, valstep=0.1)
slider_y2 = Slider(ax_y2, 'V2 Y', -5, 5, valinit=init_y2, valstep=0.1)

# Connect sliders to update function
slider_x1.on_changed(update)
slider_y1.on_changed(update)
slider_x2.on_changed(update)
slider_y2.on_changed(update)

# Initial plot
update(None)

# Add explanation text
explanation = """
COSINE SIMILARITY EXPLAINED:
Cosine similarity measures the cosine of the angle between two vectors.
• Range: -1 (opposite) to +1 (identical direction)
• Formula: cos(θ) = (A·B) / (||A|| ||B||)
• Use sliders to adjust vectors and see how similarity changes
"""
fig.text(0.5, 0.01, explanation, ha='center', fontsize=9, 
         bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))

plt.show()
```
