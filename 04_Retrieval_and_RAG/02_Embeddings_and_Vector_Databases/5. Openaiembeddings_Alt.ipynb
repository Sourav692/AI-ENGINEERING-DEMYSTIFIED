## Step 2: Creating Embeddings

### Single Text Embedding

Use `embed_query()` for embedding a single piece of text (like a user's search query).

**When to use embed_query():**
- Embedding a user's search query
- Embedding a single document
- When you need one embedding at a time

```python
# EMBEDDING A SINGLE TEXT
# ========================
# embed_query() converts one text string into a vector

single_text = "Langchain and Rag are amazing frameworks and projects to work on"

# Call the OpenAI API to get the embedding
# This makes an API call and returns a list of floats
single_embeddings = embeddings.embed_query(single_text)

# Examine the result
print(f"Input text: '{single_text}'")
print(f"Output dimensions: {len(single_embeddings)}")  # Should be 1536
print(f"\nFirst 10 values of the embedding vector:")
print(single_embeddings[:10])
```

### Multiple Text Embeddings (Batch Processing)

Use `embed_documents()` for embedding multiple texts at once. This is more efficient than calling `embed_query()` multiple times.

**When to use embed_documents():**
- Embedding a collection of documents for indexing
- Building a vector database
- Processing documents in bulk

**Efficiency Tip**: Batch processing reduces API calls and is faster!

## Step 3: Measuring Similarity with Cosine Similarity

### Understanding Cosine Similarity

Cosine similarity measures the angle between two vectors, regardless of their magnitude. It's the standard metric for comparing embeddings.

**Formula**: cos(θ) = (A · B) / (||A|| × ||B||)

**Interpretation**:
- **1.0**: Identical meaning
- **0.7 - 0.9**: Very similar
- **0.4 - 0.7**: Somewhat related  
- **< 0.4**: Different meanings

This is how semantic search and RAG systems find relevant documents!

```python
# Example 1: Finding similar sentences
sentences = [
    "The cat sat on the mat",
    "A feline rested on the rug",
    "The dog played in the yard",
    "I love programming in Python",
    "Python is my favorite programming language"
]
```

```python
# IMPLEMENTING COSINE SIMILARITY
# ===============================
# This is the core function for comparing embeddings

import numpy as np

def cosine_similarity(vec1, vec2):
    """
    Calculate the cosine similarity between two vectors.
    
    Cosine similarity measures the angle between two vectors:
    - Result close to 1.0: Very similar (small angle)
    - Result close to 0.0: Not related (perpendicular)
    - Result close to -1.0: Opposite meanings (opposite direction)
    
    Args:
        vec1: First embedding vector (list or numpy array)
        vec2: Second embedding vector (list or numpy array)
    
    Returns:
        float: Similarity score between -1 and 1
    """
    # Step 1: Calculate the dot product of the two vectors
    # This measures how much the vectors point in the same direction
    dot_product = np.dot(vec1, vec2)
    
    # Step 2: Calculate the magnitude (length) of each vector
    norm_a = np.linalg.norm(vec1)  # ||A||
    norm_b = np.linalg.norm(vec2)  # ||B||
    
    # Step 3: Divide dot product by product of magnitudes
    # This normalizes the result to be between -1 and 1
    return dot_product / (norm_a * norm_b)

print("Cosine similarity function defined!")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings=OpenAIEmbeddings(model="text-embedding-3-small")
embeddings
```

```python
sentence_embeddings=embeddings.embed_documents(sentences)
```

```python
## Calculate and display the similarity between all pairs in tabular format

import pandas as pd

data = []
for i in range(len(sentences)):
    for j in range(i+1, len(sentences)):
        similarity = cosine_similarity(sentence_embeddings[i], sentence_embeddings[j])
        data.append({
            "Sentence 1": sentences[i],
            "Sentence 2": sentences[j],
            "Cosine Similarity": round(similarity, 3)
        })

df = pd.DataFrame(data)
display(df)  # or print(df) if not in a notebook
```

```python
# ==============================================================================
# STEP 4: SEMANTIC SEARCH - A PRACTICAL APPLICATION
# ==============================================================================
# This is the foundation of RAG (Retrieval-Augmented Generation)!
# We'll find documents that are semantically similar to a query,
# even if they don't share the same keywords.

# Our document collection - imagine this as a knowledge base
documents = [
    "LangChain is a framework for developing applications powered by language models",
    "Python is a high-level programming language",
    "Machine learning is a subset of artificial intelligence",
    "Embeddings convert text into numerical vectors",
    "The weather today is sunny and warm"
]

# The user's search query
# Notice: "What is Langchain?" should match the first document
# even though the exact words are different!
query = "What is Langchain?"

print(f"Query: '{query}'")
print(f"Number of documents to search: {len(documents)}")
```

```python
def semantic_search(query, documents, embeddings_model, top_k=3):
    """
    Simple semantic search implementation that returns results as a DataFrame.
    
    This function demonstrates the core concept behind RAG retrieval:
    1. Convert the query to an embedding
    2. Convert all documents to embeddings
    3. Find documents with highest similarity to the query
    4. Output as a pandas DataFrame
    
    Args:
        query: The search query string
        documents: List of document strings to search through
        embeddings_model: The embedding model to use
        top_k: Number of top results to return
    
    Returns:
        Pandas DataFrame with the top_k most similar documents and their scores
    """
    import pandas as pd

    # STEP 1: Embed the query
    query_embedding = embeddings_model.embed_query(query)
    
    # STEP 2: Embed all documents
    doc_embeddings = embeddings_model.embed_documents(documents)
    
    # STEP 3: Calculate similarity between query and each document
    data = []
    for i, doc_emb in enumerate(doc_embeddings):
        similarity = cosine_similarity(query_embedding, doc_emb)
        data.append({
            "Score": similarity,
            "Document": documents[i]
        })
    
    # STEP 4: Sort by similarity (highest first) and return top_k results as a DataFrame
    df = pd.DataFrame(data)
    df = df.sort_values(by="Score", ascending=False).head(top_k).reset_index(drop=True)
    return df

print("Semantic search function defined!")
```

```python
results=semantic_search(query,documents,embeddings)
results
```
