### `HandoffState`

```python
# ============ HANDOFFSTATE ================================================class HandoffState(TypedDict):    messages: Annotated[list[BaseMessage], add_messages]    current_agent: str    handoff_reason: str    context_summary: str
```

### `HandoffDecision`

```python
# ============ HANDOFFDECISION =============================================class HandoffDecision(BaseModel):    handoff_to: Literal["sales", "support", "billing", "stay", "end"] = Field(        description="Which agent to hand off to"    )    reason: str = Field(description="Reason for handoff")    context: str = Field(description="Key context to pass to next agent")
```
